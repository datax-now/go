```python
%%js
function renderDragDropBoard({ model, el }) {
  const style = document.createElement("style");
  style.textContent = `
    .task-board { width: min(100%, 720px); box-sizing: border-box; padding: 22px; border: 1px solid #dbe3ef; border-radius: 22px; background: linear-gradient(145deg, #fff, #f8fafc); box-shadow: 0 18px 44px rgba(15,23,42,.11); color: #172033; font-family: Inter, ui-sans-serif, system-ui, sans-serif; }
    .task-board * { box-sizing: border-box; }
    .task-board__header { display: flex; align-items: flex-end; justify-content: space-between; gap: 16px; margin-bottom: 17px; }
    .task-board__eyebrow { margin: 0 0 3px; color: #ea580c; font-size: 11px; font-weight: 850; letter-spacing: .14em; text-transform: uppercase; }
    .task-board h3 { margin: 0; font-size: 23px; letter-spacing: -.035em; }
    .task-board__add { display: flex; gap: 7px; }
    .task-board__add input { min-width: 0; width: 180px; border: 1px solid #cbd5e1; border-radius: 10px; padding: 9px 10px; font: 12px inherit; }
    .task-board__add button { border: 0; border-radius: 10px; padding: 9px 12px; background: #172033; color: white; cursor: pointer; font: 750 11px/1 inherit; }
    .task-board__lanes { display: grid; grid-template-columns: repeat(3, 1fr); gap: 11px; }
    .task-board__lane { min-height: 210px; padding: 11px; border: 1px solid #e2e8f0; border-radius: 15px; background: #f8fafc; transition: border-color 140ms ease, background 140ms ease; }
    .task-board__lane.is-over { border-color: #f97316; background: #fff7ed; }
    .task-board__lane-title { display: flex; justify-content: space-between; margin: 0 2px 10px; color: #64748b; font-size: 11px; font-weight: 850; letter-spacing: .08em; text-transform: uppercase; }
    .task-board__badge { display: grid; width: 19px; height: 19px; place-items: center; border-radius: 7px; background: #e2e8f0; color: #475569; font-size: 10px; }
    .task-board__list { display: grid; gap: 8px; }
    .task-board__card { padding: 11px; border: 1px solid #e2e8f0; border-left: 3px solid #f97316; border-radius: 11px; background: white; box-shadow: 0 3px 8px rgba(15,23,42,.06); cursor: grab; font-size: 12px; font-weight: 700; transition: transform 140ms ease, opacity 140ms ease; }
    .task-board__card:hover { transform: translateY(-1px); }
    .task-board__card.is-dragging { opacity: .42; }
    .task-board__card small { display: block; margin-top: 6px; color: #94a3b8; font-size: 9px; font-weight: 750; text-transform: uppercase; }
    .task-board__hint { margin: 13px 0 0; color: #64748b; font-size: 10px; text-align: center; }
    @media (max-width: 580px) { .task-board__header { align-items: stretch; flex-direction: column; } .task-board__add input { width: 100%; } .task-board__lanes { grid-template-columns: 1fr; } .task-board__lane { min-height: 90px; } }
  `;

  const board = document.createElement("section");
  board.className = "task-board";
  board.innerHTML = `
    <header class="task-board__header"><div><p class="task-board__eyebrow">HTML Drag and Drop</p><h3>Priority Flow Board</h3></div><form class="task-board__add"><input aria-label="New task" placeholder="Add a task..." maxlength="42" /><button>Add</button></form></header>
    <div class="task-board__lanes"></div><p class="task-board__hint">Drag cards between lanes, or click a card to move it forward.</p>
  `;
  el.replaceChildren(style, board);

  const lanes = [{ id: "backlog", label: "Backlog" }, { id: "doing", label: "In progress" }, { id: "done", label: "Complete" }];
  const lanesRoot = board.querySelector(".task-board__lanes");
  const form = board.querySelector("form");
  const input = board.querySelector("input");
  let draggedId = null;

  function moveTask(id, lane) {
    const tasks = model.get("tasks").map((task) => task.id === id ? { ...task, lane } : task);
    model.set("tasks", tasks);
    model.save_changes();
  }

  function render() {
    const tasks = model.get("tasks");
    lanesRoot.replaceChildren();
    lanes.forEach((lane, laneIndex) => {
      const laneElement = document.createElement("section");
      laneElement.className = "task-board__lane";
      laneElement.dataset.lane = lane.id;
      const laneTasks = tasks.filter((task) => task.lane === lane.id);
      laneElement.innerHTML = `<h4 class="task-board__lane-title">${lane.label}<span class="task-board__badge">${laneTasks.length}</span></h4><div class="task-board__list"></div>`;
      const list = laneElement.querySelector(".task-board__list");
      laneTasks.forEach((task) => {
        const card = document.createElement("article");
        card.className = "task-board__card";
        card.draggable = true;
        card.tabIndex = 0;
        card.innerHTML = `${task.title}<small>Click to advance</small>`;
        card.addEventListener("dragstart", () => { draggedId = task.id; card.classList.add("is-dragging"); });
        card.addEventListener("dragend", () => { draggedId = null; card.classList.remove("is-dragging"); });
        card.addEventListener("click", () => moveTask(task.id, lanes[Math.min(laneIndex + 1, lanes.length - 1)].id));
        card.addEventListener("keydown", (event) => { if (event.key === "Enter" || event.key === " ") { event.preventDefault(); card.click(); } });
        list.appendChild(card);
      });
      laneElement.addEventListener("dragover", (event) => { event.preventDefault(); laneElement.classList.add("is-over"); });
      laneElement.addEventListener("dragleave", () => laneElement.classList.remove("is-over"));
      laneElement.addEventListener("drop", (event) => { event.preventDefault(); laneElement.classList.remove("is-over"); if (draggedId) moveTask(draggedId, lane.id); });
      lanesRoot.appendChild(laneElement);
    });
  }

  form.addEventListener("submit", (event) => {
    event.preventDefault();
    const title = input.value.trim();
    if (!title) return;
    model.set("tasks", [...model.get("tasks"), { id: `task-${Date.now()}`, title, lane: "backlog" }]);
    model.save_changes();
    input.value = "";
  });

  model.on("change:tasks", render);
  render();
  return () => model.off("change:tasks", render);
}

display_widget(datax.widget({
  model: { tasks: [{ id: "task-1", title: "Explore browser events", lane: "backlog" }, { id: "task-2", title: "Build an interactive widget", lane: "doing" }, { id: "task-3", title: "Share synchronized state", lane: "done" }] },
  createView: renderDragDropBoard
}));
```
