# Auto-Implementation Demo

This notebook demonstrates the DataX auto-implementation flow: when a cell contains a recognized placeholder **{}**, the kernel asks the AI implementation service for code, updates the cell source, and stops before executing either the original placeholder or the generated code.

Run a placeholder cell then the auto implementating will be triggered. After a placeholder cell is updated, review the generated code, then run the updated cell again if you want to execute it.

> Requirement: AI features need a GenAI is configured in Jupyterlite.

## 1. Check the Feature Wiring

This cell does not trigger generation. It only confirms that the browser runtime exposes the auto-implementation bridge.

```python
%%js
const checks = {
  hasModule: typeof Module !== 'undefined',
  implementCodeType: typeof Module?.implementCode,
  autoImplementationEnabled: globalThis.dataxNow?.ai?.autoImplementationEnabled,
  autoImplementationTimeoutMs: globalThis.dataxNow?.ai?.autoImplementationTimeoutMs,
};

console.log('Auto-implementation bridge:', checks.implementCodeType === 'function' ? 'ready' : 'not available');
console.log(JSON.stringify(checks, null, 2));
```

## 7. What to Look For

Each placeholder cell should finish with an auto-implementation review result instead of executing. The cell source should be replaced with suggested code. Generated code stays visible in the notebook, and the user decides whether to run it.

The intended classification is:

| Section | Language | Expected pattern |
| --- | --- | --- |
| Algorithmic | Python | Deterministic Python implementation |
| Semantic | Python | `datax.typedagent()` wrapper or equivalent structured semantic call |
| Hybrid | JavaScript | Deterministic outer pipeline plus `datax.typedagent()` sub-steps |
| Explicit marker | R | Replacement generated from the marker instruction |
