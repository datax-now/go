# Quick Start: Building Multi-Language Data Apps

This notebook introduces DataX.now's language-aware cells. Run the examples in order to see how Python, R, JavaScript, TypeScript, and executable Markdown work together.

## 1. Python (Default)

Cells use Python by default, so you can write and run standard Python without a magic command.

````python
# Standard Python — works exactly like any Python notebook
print("Hello from Python!")

x = [1, 2, 3, 4, 5]
print(f"Sum: {sum(x)}, Mean: {sum(x)/len(x)}")
````

````python
import math
result = math.sqrt(144)
print(f"√144 = {result}")
````

````python
# Python visualization: matplotlib is the standard plotting stack in notebooks
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 300)
fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(x, np.sin(x), label="sin(x)", color="#1f77b4", linewidth=2.5)
ax.plot(x, np.cos(x), label="cos(x)", color="#ff7f0e", linewidth=2.5)
ax.fill_between(x, np.sin(x), np.cos(x), where=np.sin(x) > np.cos(x), color="#2ca02c", alpha=0.18)
ax.set_title("Matplotlib: Python-native scientific plotting")
ax.set_xlabel("radians")
ax.set_ylabel("value")
ax.grid(alpha=0.25)
ax.legend()
plt.show()
````

## 2. R

Start a cell with `%%R` (or `%R`) to run R code in the R interpreter.

````python
%%R
# R code runs in a real R interpreter
message("Hello from R!")

data <- c(10, 20, 30, 40, 50)
cat("R mean:", mean(data), "
")
cat("R sd:",   sd(data),   "
")
````

## 3. JavaScript

Start a cell with `%%js` to run JavaScript.

````python
%%js
// JavaScript runs in a real JS engine
console.log("Hello from JavaScript!");

const nums = [1, 2, 3, 4, 5];
const sum = nums.reduce((a, b) => a + b, 0);
console.log();
````

````python
%%js
// JavaScript runs in the kernel worker; use console output for portable feedback.
const message = `Hello from JavaScript at ${new Date().toLocaleTimeString()}`;
console.log(message);

await new Promise(resolve => setTimeout(resolve, 250));
console.log("JavaScript async/await completed inside the xeus-x kernel.");
````

## 4. TypeScript

Start a cell with `%%ts` (or `%%typescript`) to run TypeScript. DataX.now compiles the code automatically before executing it.

````python
%%ts
// TypeScript with full type support
function greet(name: string): string {
    return name ;
}

const languages: string[] = ["Python", "R", "JavaScript", "TypeScript"];
languages.forEach(lang => console.log(greet(lang)));
````

## 5. Executable Code Blocks in Markdown Cells

Start a cell with `%%md` to render Markdown with fenced code blocks that execute as part of the cell.

````python
%%md
#### Mixed Markdown + Code

A single Markdown cell can render prose and execute fenced code blocks in multiple languages.

```python
import math
values = [1, 2, 3, 4, 5]
print("Python sqrt(144):", math.sqrt(144))
print("Python mean:", sum(values) / len(values))
```

```r
data <- c(10, 20, 30, 40, 50)
cat("R mean:", mean(data), "\n")
cat("R sd:", sd(data), "\n")
```

```js
const nums = [1, 2, 3, 4, 5];
const total = nums.reduce((a, b) => a + b, 0);
console.log(`\nJavaScript sum: ${total}`);
```
````

## 6. Language Switching Summary

| Magic | Language | Example |
|-------|----------|---------|
| *(none)* | Python | `print("Hello from Python!")` |
| `%%R` or `%R` | R | `mean(c(10, 20, 30))` |
| `%%js` | JavaScript | `console.log("Hello from JavaScript!")` |
| `%%ts` | TypeScript | `const language: string = "TypeScript"` |
| `%%md` | Markdown+code | fenced `python`, `r`, and `js` blocks |

These are the only commands you need to switch languages. See the other notebooks for more advanced features.
