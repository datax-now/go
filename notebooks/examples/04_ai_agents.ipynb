# AI Agents & Smart Commands

DataX Kernel includes built-in AI agent commands that let you generate code,
ask questions, and run cross-language tasks — all from within a notebook cell.

## 1. `@ask` — Ask a Question

`@ask` answers natural-language questions about your data or code
without modifying any state.

```python
topic = "light"
```

```python
@ask what is {topic}?
```

## 2. `@pygo` — Generate and Run Python Code

`@pygo` generates Python code from a natural-language description and immediately executes it in the Python sub-kernel.

```python
@pygo Create a pandas DataFrame with columns "product", "units_sold",
and "revenue" for five fictional products and display it.
```

## 3. `@rgo` — Generate and Run R Code

`@rgo` targets the R sub-kernel.

```python
@rgo Please say "hi"
```

## 4. `@jsgo` — Generate and Run JavaScript

`@jsgo` targets the JavaScript sub-kernel.

```python
@jsgo Please display a welcome message
```

## 5. Auto-Fix (`@autofix`)

If a cell raises an error, `@autofix` analyses the traceback and
suggests (or applies) a corrected version automatically.

```python
# assert_error
# This cell has a deliberate bug
if True:
print("Incorrectly indented")
```

## Summary

| Command | Target | Description |
|---------|--------|-------------|
| `@ask` | — | Answer a question (no execution) |
| `@pygo` | Python | Generate + run Python |
| `@jsgo` | JavaScript | Generate + run JS |
| `@rgo` | R | Generate + run R |
| `@autofix` | last error | Fix the previous cell's error |

All commands support multi-line descriptions — just keep writing on
subsequent lines without an empty line break.
