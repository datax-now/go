```python
import altair as alt
import pandas as pd
```

```python
# Official-style bar chart from a pandas DataFrame
data = pd.DataFrame({
    "x": ["A", "B", "C", "D", "E"],
    "y": [5, 3, 6, 7, 2]
})

alt.Chart(data).mark_bar().encode(
    x="x:N",
    y="y:Q",
    color="x:N"
).properties(width=500, height=300, title="Simple Bar Chart")
```

```python
# Scatter example with explicit encodings
cars = pd.DataFrame({
    "Horsepower": [130, 165, 150, 140, 198, 220, 215, 225],
    "Miles_per_Gallon": [18, 15, 18, 16, 15, 14, 14, 14],
    "Origin": ["USA", "USA", "USA", "USA", "USA", "USA", "USA", "USA"]
})

alt.Chart(cars).mark_circle(size=90).encode(
    x="Horsepower:Q",
    y="Miles_per_Gallon:Q",
    color="Origin:N",
    tooltip=["Horsepower", "Miles_per_Gallon", "Origin"]
).properties(width=500, height=300, title="Horsepower vs MPG")
```

```python
# Add interactivity (pan/zoom/selection) with one method call
base = pd.DataFrame({
    "a": list("ABCDEFGHI"),
    "b": [28, 55, 43, 91, 81, 53, 19, 87, 52]
})

chart = alt.Chart(base).mark_bar().encode(
    x="a:O",
    y="b:Q",
    color="b:Q"
)

chart.interactive()
```
