```python
from bokeh.io import output_notebook, show
from bokeh.plotting import figure

# Render plots inline in the notebook
output_notebook()
```

```python
# Official-style first plot: add circle glyphs to a figure
p = figure(width=420, height=300, tools="pan,wheel_zoom,box_zoom,reset,save", title="First Bokeh Plot")
p.scatter([1, 2.5, 3, 2], [2, 3, 1, 1.5], size=14, alpha=0.6, color="#f97316")
show(p)
```

```python
# Multi-line example inspired by the first-steps tutorial
x = [1, 2, 3, 4, 5]
y1 = [6, 7, 2, 4, 5]
y2 = [2, 3, 4, 5, 6]

p2 = figure(title="Multiple Line Example", x_axis_label="x", y_axis_label="y", width=600, height=320)
p2.line(x, y1, legend_label="Series A", color="#2563eb", line_width=2)
p2.line(x, y2, legend_label="Series B", color="#dc2626", line_width=2)
p2.legend.location = "top_left"
show(p2)
```

```python
from bokeh.models import ColumnDataSource, HoverTool

# A small data-source example with hover tooltips
source = ColumnDataSource({
    "x": [1, 2, 3, 4, 5],
    "y": [5, 3, 4, 8, 7],
    "label": ["A", "B", "C", "D", "E"]
})

p3 = figure(title="ColumnDataSource + Hover", width=600, height=320, tools="pan,wheel_zoom,reset,save")
p3.scatter("x", "y", size=14, source=source, color="#10b981")
p3.add_tools(HoverTool(tooltips=[("Label", "@label"), ("x", "@x"), ("y", "@y")]))
show(p3)
```
