# Install DataX.now 🚀

Follow the illustrated steps below to install **DataX.now** and get started in just a few moments.

![Installation steps](../data/install.png)

## Use DataX.now Like a Local App 🖥️

Once installation is complete, DataX.now appears on your desktop like a regular application. For quick access, you can also pin it to your taskbar.

![DataX.now desktop icon](../data/local_icon.png)

## Launch DataX.now as a Local Application ▶️

Open DataX.now from your desktop or taskbar. It launches as a standalone application, without browser tabs or other browser interface elements.

![DataX.now launched as a local application](../data/installed_app.png)


## Run Offline: Yes! 📴

As you work in DataX.now, the components required by your notebooks are cached automatically.

Once the required components have been cached, you can continue using the application offline.

⚠️ A remote GenAI model still requires an internet connection.
