# 🤖 Configure GenAI providers

GenAI settings are managed through the frontend extension. Once a provider is active, its model, base URL, request headers, and API key are detected automatically—so there is no need to edit the notebook configuration manually.

# 🧰 Built-in GenAI provider

For convenience, a lightweight GenAI provider is configured by default. However, high traffic can make the default connection unstable.

For better performance, reliability, and privacy, we recommend configuring your own GenAI provider.

# 🔑 No API key?

Agnes, a global top-10 AI lab, currently provides free access to frontier models. You can request an API key at [platform.agnes-ai.com](https://platform.agnes-ai.com/login), then configure it by following the [Agnes setup guide](https://agnes-ai.com/zh-Hans/docs/cid7).

# 🚀 Usage

Most AI providers require an API key to access their models. Keep your key private and follow the provider's documentation to create one.

The setup steps vary by provider. For provider-specific instructions, refer to the documentation for the service you want to use.

## Add a provider with an API key

For providers such as Anthropic, Mistral AI, or OpenAI:

1. Open **AI Settings**.
2. Select **Add a new provider**.
3. Enter the provider details, including your API key.
4. Open the chat and select the provider you just added.

![Dialog for adding a new provider](../data/Agnes.png)

💡 If a provider is not listed, check its documentation for the required base URL, model name, and authentication headers.
